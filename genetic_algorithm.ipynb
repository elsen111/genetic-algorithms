```python
#  importing  required  libraries
import numpy as np 
import pandas as pd 
import math
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LogisticRegression
```

```python
# reading  the  dataset
data = pd.read_csv("breast-cancer.csv")
feature_list=data.iloc[:300,1:10]
target=data.iloc[:300,-1]
```

```python
print('Feature List\n',feature_list, '\n\nTarget = ',target)
```

```python
# creating  initial  population
def init_population(m,n):
    return np.array([[math.ceil(e) for e in pop] for pop in (np.random.rand(m,n)-0.5)])
d=init_population(150,9)
print("\n\nInitial  population :\n",d)
print("\n\nSet  of  chromosome :\n")
for chromosome in d:
    print(chromosome) #### individual or chromosome
```

```python
x=feature_list
y=target
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=10)
log_model=LogisticRegression()

# creating fitness function
def fitness_score(population):
    scores = []
    for chromosome in population:
        log_model.fit(x_train.iloc[:,chromosome],y_train)
        predictions = log_model.predict(x_test.iloc[:,chromosome])
        scores.append(accuracy_score(y_test,predictions))
    scores, population = np.array(scores), np.array(population) 
    inds = np.argsort(scores)
    return scores[inds][::-1],list(population[inds,:][::-1])
fit=fitness_score(d)

print(fit)
```

```python
def single_point_crossover(population):
    r,m, n = population.shape[0], population.shape[1], np.random.randint(1,population.shape[1])         
    for i in range(0,r,2):                
        population[i], population[i+1] = np.append(population[i][0:n],population[i+1][n:m]),np.append(population[i+1][0:n],population[i][n:m])        
    return population


print(single_point_crossover(d))
```

```python
def flip_mutation(population):
    return population.max() - population
```

```python
print(flip_mutation(d))
```

```python
def random_selection(population):
    r = population.shape[0]
    new_population = population.copy()    
    for i in range(r):        
        new_population[i] = population[np.random.randint(0,r)]
    return new_population
```

```python
print(random_selection(d))
```
