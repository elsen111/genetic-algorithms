```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# reading the dataset
data = pd.read_csv('breast-cancer.csv', nrows = 300)
```

```python
#array = dataframe.values
x = data.iloc[:,1:10].values
y = data.iloc[:,10].values
```

```python
#changing the values of last column
y[y==2] = 0
y[y==4] = 1
```

```python
# Spliting the data into two parts : training and testing
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.4, random_state = 0)
```

```python
omeqa=np.zeros([9,1])
omeqa0=0.1
iterations = 50000
alpha=0.005
```

```python
def sigmoid(omeqa,omeqa0,x):
    z=np.dot(omeqa.T,x)+omeqa0
    a=1/(1+np.exp(-z))
    return a
```

```python
def cost(omeqa,x,y):
    a=sigmoid(omeqa,omeqa0,x)
    l=-(np.dot(np.log(a),y)+np.dot(np.log(1-a),1-y))
    l/=train_length
    return a,l
```

```python
def gradient(omeqa,x,y):
    a,l=cost(omeqa, x, y)
    domeqa=np.dot(x,a.T-y)
    domeqa/=train_length
    domeqa0=np.sum(a.T-y)
    domeqa0/=train_length
    return domeqa, domeqa0
```

```python
a,l=cost(omeqa, x_train, y_train)
```

```python
for i in range(iterations):
    a,l=cost(omeqa, x_train, y_train)
    domeqa,domeqa0=gradient(omeqa, x_train, y_train)
    omeqa=omeqa-alpha*domeqa
    omeqa0=omeqa0-alpha*domeqa0
```

```python
def predict(omeqa , omeqa0 ,x):
    y_pred=sigmoid(omeqa, omeqa0 , x)
    for i in y_pred:
        for j in range(len(i)):
            if y_pred[0][j]>0.5:
                y_pred[0][j]=1
            else:
                y_pred[0][j]=0
            
    return y_pred 
```
